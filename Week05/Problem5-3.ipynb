```python
import pandas as pd
import statsmodels.api as sm

data = pd.read_csv("SpaceShuttle.csv")

# Remove the row with missing data
data = data.dropna(subset=["Temperature", "nFailures"])

# r = number of failed O-rings
# m = 6 O-rings per shuttle
data["Success"] = 6 - data["nFailures"]

# Binomial response: [successes, failures]
y = data[["Success", "nFailures"]]

# Predictor: temperature
X = sm.add_constant(data[["Temperature"]])

# Fit model
model = sm.GLM(
    y,
    X,
    family=sm.families.Binomial()
)

result = model.fit()

result.summary()
```
